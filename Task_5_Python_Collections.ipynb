```python
# LIST
numbers = [40, 20, 10, 30]

numbers.append(50)
print("After append:", numbers)

numbers.remove(20)
print("After remove:", numbers)

numbers.sort()
print("After sort:", numbers)


# TUPLE
numbers_tuple = (10, 20, 30, 20)

print("Tuple:", numbers_tuple)
print("Count of 20:", numbers_tuple.count(20))
print("Index of 30:", numbers_tuple.index(30))


# SET
numbers_set = {10, 20, 30, 20, 10}

print("Set:", numbers_set)

numbers_set.add(40)
print("After add:", numbers_set)

numbers_set.remove(20)
print("After remove:", numbers_set)


# DICTIONARY
student = {
    "Name": "Anu",
    "Age": 19,
    "Marks": 85
}

print("Dictionary:", student)

student["Marks"] = 90
print("After update:", student)

student["Grade"] = "A"
print("After adding Grade:", student)
```

```python
# ============================================
# FUNCTIONS FOR REUSABLE DATA TASKS
# Data Science Internship - Veda Technology
# ============================================

import pandas as pd


# ============================================
# 1. CREATE STUDENT DATASET
# ============================================

data = {
    "Name": ["Anu", "Ravi", "Priya", "Kiran", "Sneha"],
    "Maths": [85, 78, 92, 67, 88],
    "Science": [90, 82, 89, 72, 91],
    "English": [80, 75, 95, 70, 85]
}

df = pd.DataFrame(data)

print("STUDENT DATASET")
print(df)


# ============================================
# 2. FUNCTION TO CALCULATE MEAN
# ============================================

def calculate_mean(values):
    return sum(values) / len(values)


# ============================================
# 3. FUNCTION TO FIND MAXIMUM
# ============================================

def find_maximum(values):
    return max(values)


# ============================================
# 4. FUNCTION TO FIND MINIMUM
# ============================================

def find_minimum(values):
    return min(values)


# ============================================
# 5. FUNCTION TO CALCULATE PERCENTAGE
# ============================================

def calculate_percentage(marks, total_marks):
    return (sum(marks) / total_marks) * 100


# ============================================
# 6. FUNCTION TO CALCULATE GRADE
# ============================================

def calculate_grade(percentage):

    if percentage >= 90:
        return "A+"

    elif percentage >= 80:
        return "A"

    elif percentage >= 70:
        return "B"

    elif percentage >= 60:
        return "C"

    elif percentage >= 50:
        return "D"

    else:
        return "F"


# ============================================
# 7. CALCULATE TOTAL MARKS
# ============================================

df["Total"] = (
    df["Maths"] +
    df["Science"] +
    df["English"]
)


# ============================================
# 8. CALCULATE MEAN
# ============================================

df["Mean"] = df["Total"] / 3


# ============================================
# 9. CALCULATE PERCENTAGE
# ============================================

df["Percentage"] = (df["Total"] / 300) * 100


# ============================================
# 10. CALCULATE GRADE
# ============================================

df["Grade"] = df["Percentage"].apply(calculate_grade)


# ============================================
# 11. DISPLAY FINAL DATASET
# ============================================

print("\nFINAL STUDENT RESULTS")
print(df)


# ============================================
# 12. FIND OVERALL MAXIMUM AND MINIMUM
# ============================================

all_marks = df[["Maths", "Science", "English"]].values.flatten()

maximum_mark = find_maximum(all_marks)
minimum_mark = find_minimum(all_marks)

print("\nMaximum Mark:", maximum_mark)
print("Minimum Mark:", minimum_mark)


# ============================================
# 13. SUBJECT-WISE MEAN
# ============================================

maths_mean = calculate_mean(df["Maths"].tolist())
science_mean = calculate_mean(df["Science"].tolist())
english_mean = calculate_mean(df["English"].tolist())

print("\nSUBJECT-WISE MEAN")
print("Maths Mean:", maths_mean)
print("Science Mean:", science_mean)
print("English Mean:", english_mean)


# ============================================
# 14. FUNCTION DEMONSTRATION
# ============================================

marks = [85, 90, 80]

mean = calculate_mean(marks)
maximum = find_maximum(marks)
minimum = find_minimum(marks)
percentage = calculate_percentage(marks, 300)
grade = calculate_grade(percentage)

print("\nFUNCTION DEMONSTRATION")
print("Marks:", marks)
print("Mean:", mean)
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Percentage:", percentage)
print("Grade:", grade)


# ============================================
# 15. STUDENT LIST
# ============================================

students = df["Name"].tolist()

print("\nSTUDENT LIST")
print(students)


# ============================================
# 16. STUDENT DICTIONARY
# ============================================

student_dict = {
    "Name": "Anu",
    "Maths": 85,
    "Science": 90,
    "English": 80
}

print("\nSTUDENT DICTIONARY")
print(student_dict)

print("Student Name:", student_dict["Name"])
print("Maths Marks:", student_dict["Maths"])


# ============================================
# 17. ADD A NEW STUDENT TO LIST
# ============================================

students.append("Rahul")

print("\nUPDATED STUDENT LIST")
print(students)


# ============================================
# 18. DISPLAY DATAFRAME COLUMNS
# ============================================

print("\nDATASET COLUMNS")
print(df.columns.tolist())


# ============================================
# 19. DISPLAY UNIQUE STUDENT NAMES
# ============================================

unique_students = set(df["Name"])

print("\nUNIQUE STUDENTS")
print(unique_students)


# ============================================
# 20. CONCLUSION
# ============================================

print("\nTask completed successfully!")
print("Five reusable functions were created:")
print("1. calculate_mean()")
print("2. find_maximum()")
print("3. find_minimum()")
print("4. calculate_percentage()")
print("5. calculate_grade()")
```
